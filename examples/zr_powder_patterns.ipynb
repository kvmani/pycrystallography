# Zirconium powder diffraction walkthrough

This notebook shows how to load the bundled α/β zirconium structures, calculate their powder X-ray diffraction peak lists with pymatgen, and visualise the resulting patterns using pycrystallography's plotting helpers.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from pycrystallography.adapters.pymatgen_adapter import DiffractionData, StructureLoader
from pycrystallography.core.models import Phase, PowderPattern
from pycrystallography.plotting.powder import plot_powder_pattern
```

```python
loader = StructureLoader.from_yaml(Path("../data/registry.yaml"))
alpha_phase = Phase(name="alpha-zr", structure=loader.get("zr_alpha"))
beta_phase = Phase(name="beta-zr", structure=loader.get("zr_beta"))
calculator = DiffractionData()
alpha_data = calculator.powder_pattern(alpha_phase.structure, two_theta_range=(10, 90), min_intensity=1.0)
beta_data = calculator.powder_pattern(beta_phase.structure, two_theta_range=(10, 90), min_intensity=1.0)
alpha_pattern = PowderPattern(
    phase=alpha_phase,
    two_theta=alpha_data[0],
    intensities=alpha_data[1],
    d_spacings=alpha_data[2],
    hkls=alpha_data[3],
)
beta_pattern = PowderPattern(
    phase=beta_phase,
    two_theta=beta_data[0],
    intensities=beta_data[1],
    d_spacings=beta_data[2],
    hkls=beta_data[3],
)
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
plot_powder_pattern(alpha_pattern, ax=ax)
ax.plot(beta_pattern.two_theta, beta_pattern.intensities, color="tab:orange", lw=1.2, label="β-Zr")
ax.legend(["α-Zr", "β-Zr"], loc="best")
ax.set_ylabel("Relative intensity")
ax.set_xlabel(r"$2\	heta$ (degrees)")
ax.set_title("Zirconium powder XRD patterns")
plt.show()
```
