# Burgers OR composite diffraction demo

This notebook demonstrates how to reproduce a composite selected-area electron diffraction (SAED) pattern for a Burgers orientation relation (OR) using the tooling provided by `pycrystallography`.

The Burgers OR links the body-centred cubic (β) and hexagonal close-packed (α) phases through the parallelisms $(110)_\beta \parallel (0001)_\alpha$ and $[\bar{1}11]_\beta \parallel [11\bar{2}0]_\alpha$.

Below we construct this relationship from CIF files provided in the repository, compute the composite diffraction pattern for the β phase zone axis $[111]$, and visualise the result.

```python
from pathlib import Path

from pycrystallography.core.millerDirection import MillerDirection
from pycrystallography.core.millerPlane import MillerPlane
from pycrystallography.core.orientationRelation import OrientationRelation

data_dir = Path('data/structureData')
beta_cif = data_dir / 'Zr-Beta.cif'
alpha_cif = data_dir / 'Zr-Alpha.cif'

beta_structure, beta_lattice = OrientationRelation.getStructureFromCif(beta_cif)
alpha_structure, alpha_lattice = OrientationRelation.getStructureFromCif(alpha_cif)

burgers_planes = [
    MillerPlane(hkl=[1, 1, 0], lattice=beta_lattice),
    MillerPlane(hkl=[0, 0, 0, 1], lattice=alpha_lattice),
]
burgers_directions = [
    MillerDirection(vector=[-1, 1, 1], lattice=beta_lattice),
    MillerDirection(vector=[1, -1, 2, 0], lattice=alpha_lattice),
]

burgers_or = OrientationRelation(
    names=['β-Zr', 'α-Zr'],
    structures=[str(beta_cif), str(alpha_cif)],
    planes=burgers_planes,
    directions=burgers_directions,
    initiateVariants=True,
)

print(burgers_or.makeORString())
```

```python
zone_axis_beta = MillerDirection(vector=[1, 1, 1], lattice=beta_lattice)
patterns = burgers_or.calculateCompositeSAED(
    parentZoneAxis=zone_axis_beta,
    productId=0,
    pc=[0.0, 0.0],
    sf=1.0,
    Tol=1,
)

burgers_or.plotSaed(patterns)
```
