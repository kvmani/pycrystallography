# Orientation variant exploration
Use this notebook to inspect variant metadata derived from an OR YAML document.

```python
from pathlib import Path

from pycrystallography.adapters.pymatgen_adapter import StructureLoader
from pycrystallography.io.or_yaml import load_orientation_library

registry = StructureLoader.from_yaml(Path('../../data/registry.yaml'))
library = load_orientation_library([Path('../or_zr.yaml')])
parent = registry.get('zr_beta')
from pycrystallography.core.models import Phase
phase_parent = Phase(name='beta-zr', structure=parent)
relation, variants, spec = library.build_orientation('burgers-zr', parent_phase=phase_parent, child_phase_name='alpha-zr')
[(variant.label, variant.orientation.to_euler(degrees=True)) for variant in variants]
```
